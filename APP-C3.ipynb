```python
#Positional Argujments
#When u call a cell a function with arguments without using keywoerds
#must be in correct order

#Positional-Only Arguments
#Tou can specify that a function can only have only positional arguments
#Add(,/) after the arguments to specify Positional-Only Arguments
```

```python
def para_func(pet,name):
    print("I hvae a ",pet)
    print("My",pet+"'s name is",name)

def my_func(name,/):
    print("Hello",name)

para_func("dog","Buddy")
para_func("Buddy","dog")
my_func("Roopa")
```

```python
#variable length Arguments
def Calculate_sum(*numbers):
    total=0
    for number in numbers:
        total+=number
    return total

print(Calculate_sum(10,20))
print(Calculate_sum(10,20,30,40))
print(Calculate_sum(5,10,15,20,25))
```

```python
#Keyword arguments
#Specify the parameter name while calling a function
#you can send the arguments with the key=value syntax
```

```python
#Keyword arguments
def students(name,age,course):
    print("Name:",name)
    print("Age",age)
    print("Course",course)

students(course="MCA",name="PGB",age=22)#key word arguments
print()
```
