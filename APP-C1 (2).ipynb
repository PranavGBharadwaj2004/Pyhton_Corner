#Finding Fibonacci series using While Loop

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1
i = 0

while i < n:
    print(a, end=" ")
    a, b = b, a + b
    i += 1
```

```python
nu = [64, 34, 25, 12, 22, 11, 90]

n = len(nu)

for i in range(n-1):
    for j in range(n- i-1):
        if nu[j] > nu[j + 1]:
            nu[j], nu[j + 1] = nu[j + 1], nu[j]

print("Sorted list:", nu)
```

```python
#Check whether the set of numbers is prime or not.
start=int(input("Enter the starting point"))
end=int(input("Enter the end point"))
for num in range(start,end+1):
    if num<2:
        continue
        
    for i in range(2,num):
        if(num%i==0):
            break
    else:
        print(num)
```

```python
#Take numbers as input until 0 is entered and prits their sum
t= 0
while True:
    num = int(input("Enter number"))
    if num == 0:
        break
    t+=num
print("Sum =",t)
```

```python
#factorial of a number
num=int(input("Enter the number:"))
f=1
for i in range(1,num+1):
    f*=i
print("Factorial of",num,"is",f)
```

```python
#Print Pattern
for i in range(1,5):
    for j in range(1,i+1):
        print("*",end=" ")
    print()
```

```python
#Print Pyramid Pattern
n = int(input("Enter the number of rows:"))
for i in range(1, n + 1):
    print(" " * (n - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```
