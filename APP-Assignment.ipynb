```python
# 1. Increasing Right Triangle
for i in range(1, 6):
    print("*" * i)
```

```python
# 2. Decreasing Right Triangle
for i in range(5, 0, -1):
    print("*" * i)
```

```python
#3. Right-Aligned Increasing Triangle
for i in range(1, 6):
    print(" ".join(["*"] * i))
```

```python
# 4. Right-Aligned Decreasing Triangle
for i in range(5, 0, -1):
    print(" ".join(["*"] * i))
```

```python
# 5. Full Pyramid
for i in range(1, 6):
    print(" " * (5 - i) + "*" * (2 * i - 1))
```

```python
# 6. Inverted Pyramid
for i in range(5, 0, -1):
    print(" " * (5 - i) + "*" * (2 * i - 1))
```

```python
# 7. Diamond
for i in range(1, 6):
    print(" " * (5 - i) + "*" * (2 * i - 1))
for i in range(4, 0, -1):
    print(" " * (5 - i) + "*" * (2 * i - 1))
```

```python
# 8. Hollow Pyramid
for i in range(1, 6):
    if i == 1:
        print(" " * 4 + "*")
    elif i == 5:
        print("*" * 9)
    else:
        print(" " * (5 - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
# 9. Solid Square

for i in range(5):
    print("* " * 5)
```

```python
# 10. Hollow Square
for i in range(5):
    if i == 0 or i == 4:
        print("* " * 5)
    else:
        print("*       *")
```

```python
# 11. Number Square
for i in range(5):
    for j in range(1, 6):
        print(j, end=" ")
    print()
```

```python
# 12. Same Number Square
for i in range(1, 6):
    for j in range(5):
        print(i, end=" ")
    print()
```

```python
# 13. Increasing Number Triangle
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# 14. Decreasing Number Triangle
for i in range(5, 0, -1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# 15. Repeated Number Triangle
for i in range(1, 6):
    for j in range(i):
        print(i, end=" ")
    print()
```

```python
# 16. Reverse Number Triangle
for i in range(1, 6):
    for j in range(5, 5 - i, -1):
        print(j, end=" ")
    print()
```

```python
# 17. Continuous Number Triangle
num = 1
for i in range(1, 5):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
# 18. Number Pyramid Pattern
for i in range(1, 6):
    print(" " * (5 - i), end="")
    for j in range(1, i + 1):
        print(j, end=" ")
    for j in range(i - 1, 0, -1):
        print(j, end=" ")
    print()
```

```python
# 19. Increasing Alphabet Triangle
for i in range(1, 6):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
# 20. Repeated Alphabet Triangle
for i in range(5):
    ch = chr(65 + i)
    for j in range(i + 1):
        print(ch, end=" ")
    print()
```

```python
# 21. Continuous Alphabet Triangle
ch = 65
for i in range(1, 6):
    for j in range(i):
        print(chr(ch), end=" ")
        ch += 1
    print()
```

```python
# 22. Reverse Alphabet Triangle
for i in range(5, 0, -1):
    for j in range(i):
        print(chr(65 + j), end=" ")
    print()
```

```python
# 23. X Pattern
for i in range(5):
    for j in range(5):
        if j == i or j == 4 - i:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# 24. Plus Pattern
for i in range(5):
    for j in range(5):
        if i == 2 or j == 2:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```

```python
# 25. Hollow Diamond
n = 5
for i in range(1, n + 1):
    if i == 1:
        print(" " * (n - i) + "*")
    else:
        print(" " * (n - i) + "*" + " " * (2 * i - 3) + "*")

for i in range(n - 1, 0, -1):
    if i == 1:
        print(" " * (n - i) + "*")
    else:
        print(" " * (n - i) + "*" + " " * (2 * i - 3) + "*")
```

```python
# 26. Hourglass Pattern
for i in range(5, 0, -1):
    print(" " * (5 - i) + "*" * (2 * i - 1))
for i in range(2, 6):
    print(" " * (5 - i) + "*" * (2 * i - 1))
```

```python
# 27. Filtering Based on Condition
numbers = [12, 45, 7, 23, 56, 18, 90]
large_numbers = []
for num in numbers:
    if num > 21:
        large_numbers.append(num)

print("large_numbers =", large_numbers)
```

```python
# 28. Reversing a String with For Loop
original_str = "Python"
reversed_str = ""
for char in original_str:
    reversed_str = char + reversed_str
print("original_str =", original_str)
print("reversed_str =", reversed_str)
```
