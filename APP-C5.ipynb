```python
#Imprting a specific function
# Instead of imprting an entire module
from math import sqrt
print(sqrt(25))
```

```python
#Imprting a specific function
# Instead of imprting an entire module
from math import sqrt,factorial
print(sqrt(25))
print(factorial(5))
```

```python
# Import using alias
import math as m

# Use the alias to access math functions
result = m.sqrt(16)  # Calculate square root of 16
print(result)        # Output: 4.0

angle = m.radians(90)  # Convert 90 degrees to radians
sin_value = m.sin(angle)  # Calculate sine of 90 degrees in radians
print(sin_value)         # Output: 1.0
```
