```python
#--Variable length Kyword Argument--- **kwargs
#Allows  a function to accept any number of keyword argument
```

```python
#**details is the keyword
```

```python
def Student_details(**details):
    for key,value in details.items():
        print(key,":",value)

Student_details(name="PGB",age=22,course="MCA",semester=2)
```

```python
#Student deails using **kwargs
def employee(**details):
    for key ,value in details.items():
        print(key,":",value)

employee(name="PGB",age=22,course="MCA",semester=2,collage="Amrita")
```

```python
#Combining *args and *kwargs
#A function can accept both types
```

```python
#Student details using both *args and *kwargs
def student(*subjects,**details):
    print("Subjects:")
    for subject in subjects:
        print("-",subject)
    print("\nStudent Details:")
    for key,value in details.items():
        print(key,":",value)

student("Python","DBMS","ML",name="PGB",semester=2)
```

```python
#Differentiate betwen print() and return()
def add_print(a,b):
    print(a+b)
def add_return(a,b):
    return a+b

result=add_print(30,100)
print("Result of print:",result)

result2=add_return(10,20)
print("Result of return:",result2)

# The first function displays the result but returns None
# The second function returns the value, allowing it to be stored or used in another 
```

```python
# Return Multiple Values
def result(marks):
    total =sum(marks)
    percentage=total/len(marks)
    return total,percentage
marks=[85,90,78,88,92]
total,percentage=result(marks)

print("Total:",total)
print("Percentage:",percentage)
```

```python
# Scope of Varibales
# Scope determines where a variable can be acccessed.
```

```python
#Check the balance using Global keyword
balance=100000
def deposit(amount):
    global balance
    balance+=amount
deposit(5000)
print("Balance:",balance)
```

```python
#Recursion
# is when a function calls itself
# This has the benefit of meaning tbht u can loop through a reach a result.
# Every recursion must have 2 parts:
#    A base case- A con=dition that stops the recursion
#    A recursive case - A function calling itself with a modified argument
# Without a base case the function would call itself fprever, acusing a stack overload
```

```python
#factoprial using recursive functiion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

# Example
print(factorial(5))  # Output: 120
```
